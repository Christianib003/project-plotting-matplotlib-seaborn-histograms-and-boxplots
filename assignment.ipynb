```python
plt.hist(df["co2"], bins=[0, 100, 200, 300, 400, 500, 600, 700])
plt.xlabel("CO2")
plt.ylabel("Number of Vehicles")
plt.title("Carbon Dioxide Distribution")
plt.grid(True)
plt.show()
```

```python
plt.hist(df["cylinders"], bins=[2, 4, 6, 8, 10, 12])
plt.xlabel("Cylinders")
plt.ylabel("Number of Vehicles")
plt.title("Cylinder Distribution Among Vehicles")
plt.grid(True)
plt.show()
```

```python
plt.hist(df["highway"], bins=[10, 20, 30, 40, 50, 60])
plt.xlabel("Highway")
plt.ylabel("Frequency of Vehicles")
plt.title("Distribution of Highway Fuel Efficiency")
plt.grid(True)
plt.show()
```

```python
numerical_df = df.select_dtypes(include=[np.number])
corr_mat = numerical_df.corr()

sns.heatmap(corr_mat, cbar=True, vmin=-1, vmax=1, cmap=sns.diverging_palette(20,220,n=200))
plt.xticks(rotation=45, ha="right")
plt.show()
```


### Interpretation of the Heatmap
#### Positive Correlations

*   **Cylinders and Displacement**: There is a strong positive correlation betwen `cylinders` and `displ` which suggests that the vehicles with a higher number of cylenders tend to have larger engine displacements.

*   **Urban City, Urban Highway, and Combined Fuel Efficiency**: There is a strong correlation between these measurements  that indicate that vehicles performing well in one driving condition, e.g: `Urban City (UCity)`, tend to perform well in others, e.g: `Urban Highway (UHighway)`.

* **Fuel Economy Score and Greenhouse Gas Score**: There is a high positive correlation between the `feScore` and `ghgScore` which means that vehicles with better fuel economy tend to produce fewer greenhouse gas emissions.

#### Negative Correlations

* **Displacement and C02 Emissions**: There is a strong negative correlation suggests that the larger engines, denoted by high number of `displ`, are associated with higher `co2` emissions.

* **Cylinders and CO2 Emissions**: There is a strong negative correlation between `cylinders` and `co2` which means that vehicles with a higher number of cylenders have a high emission of co2.

* **Fuel Efficiency and CO2 Emissions**: There is a negative correlation between `UCity`, `UHighway`,`comb`, `feScore`, and `co2` which means that vehicles with better fuel economy tend to produce fewer CO2 emissions.

#### Other Observations
* **PV2 and PV4**: There no strong correlation between `pv2` and `pv4` with other characteristics which may indicate that they may not have influence on the vehicle characteristics in the dataset.
